```python
class Solution:
    def productExceptSelf(self,nums: list[int]) -> list[int]:
        '''
        Return an array where each element is the product of all elements
        in the input list except the element at the same index.

        Args:
            nums (list[int]): A list of integers.

        Returns:
            list[int]: A list where each element contains the product of
                    all elements in nums except nums[i].

        Example:
            productExceptSelf([1, 2, 3, 4])
            [24, 12, 8, 6]
        '''
        n = len(nums)
        answer = [1] * n
        
        # Step 1: Calculate prefix products
        prefix = 1
        for i in range(n):
            answer[i] = prefix
            prefix *= nums[i]
            
        # Step 2: Calculate suffix products and combine
        suffix = 1
        for i in range(n - 1, -1, -1):
            answer[i] *= suffix
            suffix *= nums[i]
        return answer
s=Solution()
# Examples
print(s.productExceptSelf([1, 2, 3, 4]))  # [24, 12, 8, 6]
print(s.productExceptSelf([-1, 1, 0, -3, 3]))  # [0, 0, 9, 0, 0]
        
```

```python
class Solution:
    def reverseWords(self,s: str) -> str:
        '''
        Reverse the order of words in a string and remove extra spaces.

        Args:
            s (str): A string containing words separated by spaces.

        Returns:
            str: A string with the words in reverse order, separated
                by a single space.

        Example:
            reverseWords("the sky is blue")
            'blue is sky the'
            reverseWords("  hello world  ")
            'world hello'
        '''
        # split() automatically handles leading, trailing, and duplicate spaces
        return " ".join(s.split()[::-1])

s=Solution()
# Examples
print(s.reverseWords("the sky is blue"))  # "blue is sky the"
print(s.reverseWords("  hello world  "))  # "world hello"
print(s.reverseWords("a good   example"))  # "example good a"
```
