```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x= np.array([1,2,3,4])
y= np.array([1.8,2.7,3.5,4.65])
```

```python
n=len(x)
```

```python
w = 0
b = 0
```

```python
eta= 0.01
epochs=100
cost_value=[]
print(f"initial W:{w:.4f}")
print(f"initial B:{b:.4f}")
print(f"initial eta:{eta:.4f}")
print(f"initial epochs:{epochs}")
```

```python
for i in range(epochs):
    y_hat= w*x+b
    error=y_hat-y

    D_w = np.mean(error*x)
    D_b = np.mean(error*1)

    w=w-eta*D_w
    b=b-eta*D_b

    y_hat= w*x+b
    cost= 0.5*np.mean((y_hat-y)**2)
    cost_value.append(cost)
    
    if (i+1) %100 == 0 or i == 0:
        print(f"Epoch {i +1}: "
              f"cost = {cost}, "
              f"w = {w}, "
              f"b ={b} ")
```

```python
plt.figure(figsize=(12,6))
plt.subplot(1,2,1)
plt.scatter(x,y)
plt.plot(y,w*x+y,'r-')
```

```python
plt.figure(figsize=(12,6))
plt.subplot(1,2,1)
plt.plot(cost_value)
```
