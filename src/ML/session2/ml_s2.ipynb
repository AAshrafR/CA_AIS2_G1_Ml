```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error,r2_score,root_mean_squared_error

import warnings
warnings.filterwarnings('ignore')
```

```python
df= pd.read_csv('Salary_Data.csv')
```

```python
df.head()
```

```python
plt.figure(figsize=(4,4))
sns.pairplot(df,x_vars='YearsExperience',
             y_vars='Salary',size=4,kind='scatter')
plt.xlabel("Years of Experience")
plt.ylabel("Salary")
plt.title("Salary predection")
plt.show()
```

```python
x = df.iloc[:,:-1]
y = df.iloc[:,1]

x_train,x_test,y_train,y_test = train_test_split(x,y,train_size=0.8,random_state=42)
```

```python
lin_reg = LinearRegression()
lin_reg.fit(x_train,y_train)
```

```python
y_hat_train = lin_reg.predict(x_train)
y_hat_train
```

```python
plt.scatter(x_train,y_train,color='red')
plt.plot(x_train,y_hat_train,color='blue')
plt.xlabel('years')
plt.ylabel('salary')
plt.title('Salary VS Years of experience')
plt.show()
```

```python
y_hat_test = lin_reg.predict(x_test)
y_hat_test
```

```python
plt.figure(figsize=(3,3))
plt.scatter(x_test,y_test,color='red')
plt.plot(x_test,y_hat_test,color='blue')
plt.xlabel('years')
plt.ylabel('salary')
plt.title('Salary VS Years of experience')
plt.show()
```

```python
mse=mean_squared_error(y_test,y_hat_test)
print(f'MSE : {mse}')
r_mse=root_mean_squared_error(y_test,y_hat_test)
print(f'RMSE : {r_mse}')
r2=root_mean_squared_error(y_test,y_hat_test)
print(f'R2 Score : {r2}')
```

```python
w=lin_reg.coef_
b=lin_reg.intercept_

inputx= eval(input('enter years of experince'))
y_pred=w*inputx+b
print(y_pred)
```
